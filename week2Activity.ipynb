```python
#create a diagram
data['Total_Sales'] = data[['NA_Sales', 'EU_Sales', 'JP_Sales']].sum(axis=1)
```

```python
idx = data.groupby('Year')['Total_Sales'].idxmax()
top_games_per_year = data.loc[idx].sort_values('Year').reset_index(drop=True)
```

```python
years = top_games_per_year['Year']
na_sales = top_games_per_year['NA_Sales']
eu_sales = top_games_per_year['EU_Sales']
jp_sales = top_games_per_year['JP_Sales']

x = np.arange(len(years))
width = 0.25
```

```python
plt.figure()

plt.bar(x - width, na_sales, width, label='NA Sales', color='blue')
plt.bar(x, eu_sales, width, label='EU Sales', color='green')
plt.bar(x + width, jp_sales, width, label='JP Sales', color='red')

plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Highest Selling Game Each Year by Region')
plt.xticks(x, years, rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
sentiment_analysis = pipeline("sentiment-analysis")
```

```python
#a function to get us the sentiment for each item in our list
def get_sentiment(text):
    result = sentiment_analysis(text[:512])[0]  # limit length
    return result["label"]

data["sentiment_label"] = data["Content"].apply(get_sentiment)

```

```python
#sentiment analysis with nltk
%pip install nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
#additional imports for nltk to run
import nltk 
nltk.download('punkt')
nltk.download('punkt_tab')      
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('vader_lexicon')
```

```python
#define stop words and lemmatizer for processing
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

#define a function to preprocess the words 
def preprocess_text(text):
    tokens = word_tokenize(text.lower())
    
    filtered_tokens = [
        token for token in tokens 
        if token.isalpha() and token not in stop_words
    ]
    
    lemmatized_tokens = [
        lemmatizer.lemmatize(token) 
        for token in filtered_tokens
    ]
    
    return " ".join(lemmatized_tokens)

```

```python
nlp_data = data
```

```python
#run the function on our dataset
nlp_data["Content"] = nlp_data["Content"].apply(preprocess_text)
```

```python
#initialise sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

#create get_sentiment function
def get_sentiment(text):
    scores = analyzer.polarity_scores(text)
    sentiment = 1 if scores['pos'] > 0 else 0

    return sentiment
```

```python
#apply sentiment analyzer function to our data

nlp_data["sentiment"] = nlp_data["Content"].apply(get_sentiment)
```
