## Part 1: Vector Database
In Part 1, we're going to begin practicing our using vector databases on text embeddings. You should already be familiar with vectors, particularly with arrays and lists. What we're doing today, is giving text a location in space to that we can see how words connect with one another. Vector databases are how many AI applications (including RAG) work. 

In your terminal, run: pip install chromadb sentence-transformers

```python
#ONLY RUN THIS ONCE
!pip install "pydantic<2.0"
```

```python
#import packages
import chromadb
from chromadb.config import Settings
from sentence_transformers import SentenceTransformer
```

```python
#initialise model
model = SentenceTransformer("all-MiniLM-L6-v2") 

client = chromadb.Client(Settings(anonymized_telemetry=False))
collection = client.create_collection("documents")
```

```python
#initialise documents
documents = [
    "Cloud computing provides scalable resources",
    "Graph databases store relationships between data",
    "Vector databases allow semantic search",
    "Machine learning models learn patterns from data"
]
```

```python
#prepare embeddings
embeddings = model.encode(documents)

collection.add(
    documents=documents,
    embeddings=embeddings.tolist(),
    ids=["1", "2", "3", "4"]
)
```

```python
#Query the database
query = "How do databases store relationships?"

query_embeddings = model.encode([query])

results = collection.query(
    query_embeddings=query_embeddings.tolist(),
    n_results=2
)

print(results["documents"])
```

## Part 2: Graph Database
In Part 2, we'll explore graph databases. These databases store relationships between entities. Whereas vector databases excel at finding "similar" items, graph databases are used to find "connected" items. These are often used for social networks, resource planning and logistics, and fraud detection. 

In your terminal run:
pip install networkx

```python
#import packages
import networkx as nx
```

```python
#define an empty graph
graph = nx.Graph()

#add people
graph.add_node("Alice")
graph.add_node("Bob")
graph.add_node("Charlie")
```

```python
#add relationships
graph.add_edge("Alice", "Bob")
graph.add_edge("Bob", "Charlie")
```

```python
#print connections
print("Nodes:", graph.nodes())
print("Connections:", graph.edges())
```

```python
#query the data
list(graph.neighbors("Alice"))
```

```python
nx.shortest_path(graph, "Alice", "Charlie")
```

### Web Scraping with Playwright
Playwright is a browser automation library that allows you to scrape dynamic websites by controlling headless browsers like Chrome, Firefox, and Edge. 

On your terminal, run:
    pip install playwright

```python
#ONLY RUN THIS ONCE
!playwright install chromium
```

```python
#import packages
from playwright.async_api import async_playwright #(we use async in jupyter to avoid errors, think about why)
import asyncio
```

```python
#open a website
async def open_site():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False)
        page = await browser.new_page()

        await page.goto("https://www.youtube.com")

        title = await page.title()
        print("Page title:", title)

        await browser.close()

    await open_site()
```

```python
#extract text
async def extract_text():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        await page.goto("https://www.vogue.com")

        heading = await page.locator("h1").inner_text()

        print("Page heading:", heading)

        await browser.close()

await extract_text()
```

This was a basic start to extracting information from dynamic websites. If you have time, try looking at other sites, try pulling out specific bits of information. See what else you can do!
